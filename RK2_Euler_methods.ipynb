```python

import numpy as np
import matplotlib.pyplot as plt

y0 = np.array([[.05, 1.5], [-0.05, 1.5], [-0.05, -1.5], [0.05, -1.7]])
t0 = -5
tend = 5
h = 0.01
def RK2(f, y0, t0, tend, h):
  t = np.arange(t0, tend, h)
  y = np.zeros([len(t), len(y0)])
  y[0] = y0


  for i in range(0,len(t)-1):
    k1 = h * f(t[i], y[i] )
    k2 = h*f(t[i] + h/2, y[i] + 1/2 * k1)
    y[i+1] = y[i] + k2
  return  t, y


def f(t, y):
  return np.array([y[0], 2*y[0] - y[1]])
#траектории в фазовой плоскости
for i in range(len(y0)):
  res,y = RK2( f, y0[i], t0, tend, h)
  plt.plot(y[:,0], y[:,1])

plt.scatter(0,0, label = "Special point", color = "pink", linewidths= 3)
plt.xlim(-1.5, 1.5)
plt.ylim(-1.5, 1.5)
plt.grid()
plt.legend()
```
