```python
import numpy as np
import matplotlib.pyplot as plt
import scipy as sp

m1, m2, m3 = 1, 1, 0.5
k1, k2, k3 = 1, 0.5, 1/3

def f(t,y):
  return np.array([y[3], y[4], y[5], (-k1*y[0] + k2*(y[1] - y[0]))/m1, (-k2*(y[1]-y[0]) + k3*(y[2] - y[1]))/m2, -k3*(y[2] - y[1])/m3])


T = 50*np.pi
N = 2**10
w = np.arange(N)*2*np.pi/T
t = np.linspace(0, T, N)
y0 = [1, 0.2, -1, 0.3, 2, 2]


res = sp.integrate.solve_ivp(f, [0,T], y0, t_eval = t)

fig, ax = plt.subplots(figsize=(12, 8))
plt.plot(t, res.y[0], color = "red", label = "x1(t)")
plt.plot(t, res.y[1], color = "purple", label = "x2(t)")
plt.plot(t, res.y[2], color = "pink", label = "x3(t)")
plt.plot(t, res.y[3], color = "red", label = "x'1(t)", linestyle= '--')
plt.plot(t, res.y[4], color = "purple", label = "x'2(t)", linestyle= '--')
plt.plot(t, res.y[5], color = "pink", label = "x'3(t)", linestyle= '--')
plt.xlabel('Время t (с)', fontsize=14)
plt.ylabel('Общее решение y(t)', fontsize=14)
plt.xlim(0,T)
plt.grid()
plt.legend()




```
